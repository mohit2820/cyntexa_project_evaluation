## Widget Creation

```python
dbutils.widgets.text("catalog" , "digital_banking")
catalog = dbutils.widgets.get("catalog")
```

## Catalog Creation For Project 

```python
%sql
CREATE CATALOG IF NOT EXISTS IDENTIFIER(:catalog);
```

## Schema For Project

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).bronze;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).silver;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).gold;
```

## Volume Creation To Store Raw CSV Files

```python
%sql
CREATE VOLUME IF NOT EXISTS IDENTIFIER(:catalog).bronze.raw_data;
```

## Creating Directories Inside Volume Raw_Data

```python
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw_data/Accounts")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw_data/Branches")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw_data/Customers")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw_data/Transaction")
```

## Table Creation 

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.bronze_customers;
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.bronze_accounts;
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.bronze_branches;
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.bronze_transactions;
```

## USING `COPY INTO` TO LOAD CSV INTO TABLES

```python
%sql
COPY INTO IDENTIFIER(:catalog).bronze.bronze_customers
FROM '/Volumes/' :catalog '/bronze/raw_data/Customers/'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS('mergeSchema' = 'true')
```

```python
%sql
COPY INTO IDENTIFIER(:catalog).bronze.bronze_accounts
FROM '/Volumes/' :catalog '/bronze/raw_data/Accounts/'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS('mergeSchema' = 'true')
```

```python
%sql
COPY INTO IDENTIFIER(:catalog).bronze.bronze_branches
FROM '/Volumes/' :catalog '/bronze/raw_data/Branches/'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS('mergeSchema' = 'true')
```

```python
%sql
COPY INTO IDENTIFIER(:catalog).bronze.bronze_transactions
FROM '/Volumes/' :catalog '/bronze/raw_data/Transactions/'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS('mergeSchema' = 'true')
```
