```python
%sql
-- KPI 1: Total Transaction Value
-- Calculate the total value of successful transactions

CREATE OR REPLACE VIEW digital_banking.gold.kpi_total_transaction_value AS
SELECT 
  SUM(amount) AS total_transaction_value
FROM 
  digital_banking.silver.silver_transactions
WHERE 
  transaction_status = 'Completed';
```

```python
%sql
SELECT * FROM digital_banking.gold.kpi_total_transaction_value
```

```python
%sql
-- KPI 2: number of successful transactions
-- Calculate the number of successful transactions

CREATE OR REPLACE VIEW digital_banking.gold.kpi_total_success_transaction_count AS
SELECT 
  COUNT(*) AS total_successful_transactions
FROM 
  digital_banking.silver.silver_transactions
WHERE 
  transaction_status = 'Completed';
```

```python
%sql
SELECT * FROM digital_banking.gold.kpi_total_success_transaction_count;
```

```python
%sql
-- KPI 3: Deposits vs Withdrawals
-- Calculate the total value of deposits and withdrawals

CREATE OR REPLACE VIEW digital_banking.gold.kpi_deposits_vs_withdrawals AS
SELECT 
  SUM(CASE WHEN transaction_type = 'Deposit' THEN amount ELSE 0 END) AS total_deposits,
  SUM(CASE WHEN transaction_type = 'Withdrawal' THEN amount ELSE 0 END) AS total_withdrawals
FROM 
  digital_banking.silver.silver_transactions
WHERE 
  transaction_status = 'Completed'
  AND transaction_type IN ('Deposit', 'Withdrawal');
```

```python
%sql
SELECT * FROM digital_banking.gold.kpi_deposits_vs_withdrawals;
```

```python
%sql
-- KPI 4: Monthly Transaction Value
-- Calculate transaction value by month to identify trends

CREATE OR REPLACE VIEW digital_banking.gold.kpi_monthly_transaction_value AS
SELECT 
  DATE_FORMAT(transaction_date, 'yyyy-MM') AS month,
  SUM(amount) AS transaction_value,
  COUNT(*) AS transaction_count
FROM 
  digital_banking.silver.silver_transactions
WHERE 
  transaction_status = 'Completed'
GROUP BY 
  DATE_FORMAT(transaction_date, 'yyyy-MM')
ORDER BY 
  month;
```

```python
%sql
SELECT * FROM digital_banking.gold.kpi_monthly_transaction_value
```
